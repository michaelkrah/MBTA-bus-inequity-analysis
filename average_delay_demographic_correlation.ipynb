```python
filename_2023_ridership = './datasets/MBTA_Bus_Ridership_by_Trip_Season_Route_Line_and_Stop/MBTA_Bus_Ridership_by_Trip_Season_Route_Line_and_Stop_Fall_2023.csv'
df_2023_ridership = pd.read_csv(filename_2023_ridership, low_memory=False)
df_2023_ridership = df_2023_ridership.groupby('route_id')['boardings'].sum().reset_index()
```

```python
print(df_2023_ridership.head())
print(avg_positive_delay_per_route.head())
print(df_demographics.head())
```

```python
# Ensures route ids are the same across datasets
def normalize_route_id(rid):
    rid = str(rid).strip()
    return rid.zfill(2) if rid.isdigit() and len(rid) == 1 else rid

df_2023_ridership['route_id'] = df_2023_ridership['route_id'].apply(normalize_route_id)
avg_positive_delay_per_route['route_id'] = avg_positive_delay_per_route['route_id'].apply(normalize_route_id)
```

```python
route_stats = pd.merge(
    df_2023_ridership,               # boardings
    avg_positive_delay_per_route,    # delay_seconds
    on='route_id',
    how='inner'                      # keep only routes that appear in both
)

# ── 2. Quick sanity-check ──
print("Merged route_stats shape:", route_stats.shape)
display(route_stats.head())    # or just print(route_stats.head()) if you're in a plain console
```

```python
rows = []
for grp, routes in reporting_group_dic.items():
    for rid in routes:
        rows.append({"reporting_group": grp,
                     "route_id": normalize_route_id(rid)})

group_map = pd.DataFrame(rows)

route_stats_grouped = pd.merge(
    route_stats,
    group_map,
    on="route_id",
    how="inner"          
)

group_summary = (
    route_stats_grouped
      .groupby("reporting_group")
      .apply(lambda g: pd.Series({
          "total_boardings": g["boardings"].sum(),
          "weighted_avg_delay": (g["delay_seconds"] * g["boardings"]).sum() / g["boardings"].sum()
      }))
      .reset_index()
)

print("Aggregated shape:", group_summary.shape)
display(group_summary.sort_values("weighted_avg_delay", ascending=False).head())
```

```python
demo = (
    df_demographics[['reporting_group', 'measure_group', 'measure',
                     'category', 'weighted_percent']]
      .copy()
)

demo = demo.merge(
    group_summary,
    on='reporting_group',
    how='inner'
)

demo['category_boardings'] = demo['total_boardings'] * demo['weighted_percent']
demo['delay_contribution'] = demo['weighted_avg_delay'] * demo['category_boardings']

delay_by_demo = (
    demo.groupby(['measure_group', 'measure', 'category'])
        .apply(lambda g: pd.Series({
            'riders_in_category': g['category_boardings'].sum(),
            'avg_delay_seconds': g['delay_contribution'].sum() / g['category_boardings'].sum()
        }))
        .reset_index()
)

delay_by_demo = delay_by_demo.sort_values('avg_delay_seconds', ascending=False)

```

```python
delay_enthnicity = (
    delay_by_demo[delay_by_demo['measure'] == 'Race']
        .sort_values('avg_delay_seconds', ascending=False)
        .reset_index(drop=True)
)

display(delay_enthnicity)
```

```python
delay_income = (
    delay_by_demo[delay_by_demo['measure'] == 'Household Income']
        .sort_values('avg_delay_seconds', ascending=False)
        .reset_index(drop=True)
)

display(delay_income)
```

```python
delay_vehicles = (
    delay_by_demo[delay_by_demo['measure_group'] == 'Vehicles']
        .sort_values('avg_delay_seconds', ascending=False)
        .reset_index(drop=True)
)

display(delay_vehicles)
```
